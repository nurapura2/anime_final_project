# 1. Basic Data Inspection

```
df = pd.read_csv('/content/anime_checkpoint_1500.csv')
print(df.head(10))
print(df.shape)
print(df.dtypes)
print(df.info())
```

```
print('num of rows and columns: ', df.shape)
print('------------------------------------------------------------------------------------')
print('columns names: ', df.columns)
print('------------------------------------------------------------------------------------')
print('data types: ', df.dtypes.to_string())
```

# 2. Missing Values

```
#detecting missing values
missing = df.isnull().sum()
missing_pct = missing / df.shape[0] * 100
missing_above_20 = missing_pct[missing_pct > 20]

print('Columns with missing values > 20%:')
print('------------------------------------------------------------------------------------')
for col, pct in missing_above_20.items():
    print(f'{col}: {missing[col]} missing ({pct:.2f}%)')
print('------------------------------------------------------------------------------------')
print(f'Total columns affected: {len(missing_above_20)}')
```
