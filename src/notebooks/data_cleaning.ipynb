# 2. Missing Values

```python
df.isnull().sum()
```

```python
#detecting missing values
missing = df.isnull().sum()
missing_pct = missing / df.shape[0] * 100
missing_above_20 = missing_pct[missing_pct > 20]

print('Columns with missing values > 20%:')
print('------------------------------------------------------------------------------------')
for col, pct in missing_above_20.items():
    print(f'{col}: {missing[col]} missing ({pct:.2f}%)')
print('------------------------------------------------------------------------------------')
print(f'Total columns affected: {len(missing_above_20)}')
```

```python
#dropping empty columns
missing_100 = missing_pct[missing_pct == 100]

for col, pct in missing_100.items():
    df = df.drop(columns=col)
    print(f'{col} has been dropped because it has 100% missing values')

df.shape
```

```python
#column has only 2 miss, so filling it with empty string
df['alt_names'] = df['alt_names'].fillna('')

#filling missing values in 'year' column with median value
df['year'] = df['year'].str.extract('(\d{4})').astype(float)  # Extracting year as a 4-digit number and converting to float
df['year'] = df['year'].fillna(df['year'].median())
df['year'] = df['year'].astype(int)

#important column for our analysis, so filling missing values with 'Unknown'
df['genres'] = df['genres'].fillna('Unknown')



#filling missing values with median value
df['director'] = df['director'].fillna('Unknown')
df['dubbing'] = df['dubbing'].fillna('Unknown')
df['studio'] = df['studio'].fillna('Unknown')

#dropping columns that are not useful for our analysis and have many missing values
df = df.drop(columns=['kinopoisk_rating', 'kinopoisk_url'])
```

# 4. Duplicate Data

```python
#checking and resolving duplicate anime_id
dup_ids = df[df['anime_id'].duplicated(keep=False)]
print(f"Rows with duplicate anime_id: {len(dup_ids)}")

#keeping only the first occurrence
df = df.drop_duplicates(subset='anime_id', keep='first')
print(f"Shape after deduplication: {df.shape}")
```

# 5. Invalid or Suspicious Values

```python
#manually checking for issues in the data
issues = {}

issues['year_out_of_range'] = df[(df['year'] < 1900) | (df['year'] > 2030)]
issues['rating_out_of_range'] = df[(df['site_rating'] < 0) | (df['site_rating'] > 10)]
issues['votes_negative'] = df[df['site_votes'] < 0]
issues['views_negative'] = df[df['site_views'] < 0]
issues['duplicate_anime_id'] = df[df['anime_id'].duplicated(keep=False)]
issues['empty_title'] = df[df['title_ru'].isna()]
issues['empty_type'] = df[df['type'].isna()]
issues['empty_status'] = df[df['status'].isna()]
issues['empty_genres'] = df[df['genres'].isna()]
```

```python
issues
```

```python
num_cols = ['year', 'site_rating', 'site_votes', 'site_views']


#z-score method
z_scores = (df[num_cols] - df[num_cols].mean()) / df[num_cols].std()
z_outliers = np.abs(z_scores) > 3

#iqr method
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1

iqr_outliers = (df[num_cols] < (Q1 - 1.5 * IQR)) | (df[num_cols] > (Q3 + 1.5 * IQR))

#merging outliers from both methods
outliers = z_outliers | iqr_outliers

#rows with issues
df_outliers = df[outliers.any(axis=1)]

#marking
df['suspicious'] = outliers.any(axis=1)

#checking
df_outliers[num_cols].sample(50)
```

We examined the outliers and suspicious values, but concluded that in this case they cannot be analyzed reliably due to the very high variance. We also believe these values are valid, as such data can realistically occur. Therefore, we decided not to remove these outliers
